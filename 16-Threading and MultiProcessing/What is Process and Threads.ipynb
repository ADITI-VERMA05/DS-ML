#### Process

A process is simply an instance of a program that is been executed.

*Requirenments*

|---------Code Segment---------|---------Data Segment----------|---------Heap---------|

|--------------------STACK--------------------|--------------------REGISTER---------------|

|---------------------------------------------------------------------------------------------|

1. **separate memory space**
One Process can't corrupt another process.
2. **I/O Requirenments**
3. Increased Execution time to switch between Processes.

Eg. 
1. Browser 
2. Excel -> process -> Id
3. App -> process

#### Thread

A Thread is a unit of execution within a process.

**Single Threaded Process**

|---------Code Segment---------|---------Data Segment----------|---------Heap---------|

|--------------------STACK--------------------|--------------------REGISTER---------------|

|---------------------------------------------(thread) ~ ------------------------------------|

Thread is having its own Stack and Register but will be sharing Code segment, Data segment and heap from the process itslef.

Eg. MS Paint -> Process ---> thread -> o (Circle creation)
---> another thread -> / (Line creation)

**Multi Threaded Process**

|---------Code Segment---------|---------Data Segment----------|---------Heap---------|

|-------STACK-------|------REGISTER--------||--------STACK-------|------REGISTER--------|

|-----------------(thread) ~ -----------------|-------------------(thread) ~ -----------------|
