#### Magic Methods

Magic methods in Python, also known as dunder methods(double underscore methods), are special methods that start and end with double underscores. These methods enable you to define the behaviour of objects for built-in operations, such as arithmetic operations, comparisions, and more.

#### Magic Methods

Magic methods are predefined methods in python that you can override to chage the behaviour of your objects. Some common magic methods include:



```python
'''
__init__ : To initialize the attributes of the class. 
__str__ : To define the string representation of the object.
__repr__ : To define the official string representation of the object.
__len__ : To define the behaviour of the built-in len() function.
__getitem__ : To define the behaviour of indexing (e.g., obj[key]).
__setitem__ : To define the behaviour of setting an item (e.g., obj[key] = value).
__delitem__ : To define the behaviour of deleting an item (e.g., del obj[key]).
'''
```

```python
## Basic Mgagic Methods

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("John", 30)
print(person)
```

```python
## Basic Mgagic Methods

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    ## overriding the default __str_ method
    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"

    ## overriding the default __repr_ method
    def __repr__(self):
        return f"Person's name is {self.name} and he is {self.age} years old"
    
person = Person("John", 30)
print(person)
print(repr(person))
```
