#### Operator Overloading

Operator overloading allows you to define the behaviour of operators(+,_,*,etc.) for custom objects. You achieve this by overriding specific magic methods in your class.

```python
##### Common Operator Overloading Magic Methods
'''
__add__ : To define the behaviour of the addition operator (+).
__sub__ : To define the behaviour of the subtraction operator (-).
__mul__ : To define the behaviour of the multiplication operator (*).
__lt__ : To define the behaviour of the less than operator (<).
__le__ : To define the behaviour of the less than or equal to operator (<=).
__eq__ : To define the behaviour of the equality operator (==).
__ne__ : To define the behaviour of the not equal operator (!=).
__gt__ : To define the behaviour of the greater than operator (>).
__ge__ : To define the behaviour of the greater than or equal to operator (>=).
__truediv__ : To define the behaviour of the true division operator (/).
__floordiv__ : To define the behaviour of the floor division operator (//).
'''
```

```python
## Mathematical operations on Vectors using operator overloading
class vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return vector(self.x * scalar, self.y * scalar)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    def __repr__(self):
        return f"vector({self.x}, {self.y})"
    
## creating vector objects
v1 = vector(2, 3)
v2 = vector(4, 5)

## using operator overloading
print(v1 + v2)  
print(v1 - v2)
print(v1 * 3)   
print(v1 == v2)
print(repr(v1))  # Official string representation
```

```python
## Overlading operators on Complex Numbers

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.imag + other.imag)

    def __sub__(self, other):
        return ComplexNumber(self.real - other.real, self.imag - other.imag)

    def __mul__(self, other):
        return ComplexNumber(self.real * other.real - self.imag * other.imag,
                             self.real * other.imag + self.imag * other.real)
        
    def __truediv__(self, other):
        denominator = other.real**2 + other.imag**2
        real_part = (self.real * other.real + self.imag * other.imag) / denominator
        imag_part = (self.imag * other.real - self.real * other.imag) / denominator
        return ComplexNumber(real_part, imag_part)
    
    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag
    
    def __repr__(self):
        return f"ComplexNumber({self.real}, {self.imag})"

    def __str__(self):
        return f"{self.real} + {self.imag}i"
    
## creating complex number objects
c1 = ComplexNumber(3, 2)
c2 = ComplexNumber(1, 7)

## using operator overloading
print(c1 + c2)  # Addition
print(c1 - c2)  # Subtraction
print(c1 * c2)  # Multiplication
print(c1 / c2)  # Division
print(c1 == c2)  # Equality check
print(repr(c1))  # Official string representation
print(str(c1))   # String representation
print(c1)        # Implicit string representation when printing
```
