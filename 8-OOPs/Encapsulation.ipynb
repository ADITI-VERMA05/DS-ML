#### Encapsulation And Abstraction

Encapsulation ans abstraction are two fundamental principles of object oriented Programming (OOP) that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while the abstraction involves hiding complex implementation details and exposoing only the necessary features.

#### Encapsulation

Encapsulation is the concept of wrapping data (variables) methods (functions) together as a single unit. It restrcts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

```python
## Encapsulation with Getter and Setter methods
'''Public, protected and private access specifiers are used to control the visibility of class members 
(attributes and methods) in object-oriented programming. They help in encapsulating the data and ensuring 
that it is accessed and modified only through defined interfaces.'''
### Public
class Person:
    def __init__(self, name, age):
        self.name = name  # Public variable
        self.age = age  # Public attribute

def get_age(self):  # Getter method 
    return self.age
        
person = Person("John", 30)
print(person.name)  # Accessing public variable 
get_age(person)  # Accessing public method
```

```python
## Private or access modifiers
class Person:
    def __init__(self, name, age):
        self.__name = name  # Private variable
        self.__age = age  # Private attribute

def get_age(self):  # Getter method 
    return self.__age
        
person = Person("Krish", 22)
print(person.__name)  # Accessing public variable 
get_age(person)  # Accessing public method
```

```python
## Protected 
class Person:
    def __init__(self, name, age, gender):
        self._name = name  # Protected variable
        self._age = age  # Protected attribute
        self._gender = gender  # Protected attribute
        
class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age,gender)
        
employee = Employee("Alice", 34, "Male")
print(employee._name)  # Accessing protected variable
```

```python
## Encapsulation with Gerter and Setter methods
class Person:
    def __init__(self, name, age):
        self.__name = name  # Private variable
        self.__age = age  # Private attribute

    def get_name(self):  # Getter method for name
        return self.__name

    def set_name(self, name):  # Setter method for name
        self.__name = name

    def get_age(self):  # Getter method for age
        return self.__age

    def set_age(self, age):  # Setter method for age
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive")
        
## Access and modify private variables using getter and setter methods
person = Person("Krish", 30)
print(person.get_name())  # Accessing name using getter
print(person.get_age())  # Accessing age using getter   
person.set_age(35)  # Setting new age using setter
print(person.get_age())  # Accessing updated age using getter   
person.set_age(-5)  # Attempting to set invalid age
# This will raise a ValueError due to the validation in the setter method
```
