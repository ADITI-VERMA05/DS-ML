#### Custom Exception Handling(Raising and Throwing an Exception)

```python
## Inheriting from Exception Class
class Error(Exception):
    """Base class for custom exceptions."""
    pass

class dobException(Error):
    pass 
```

```python
year = int(input("Enter your year of birth: "))
age = 2025 - year

try:
    if age <= 30 and age >= 18:
        print("Age is valid so you can apply for the exam.")
    else:
        raise dobException
except dobException:
    print("Sorry, Age is not valid so you cannot apply for the exam.")
```
