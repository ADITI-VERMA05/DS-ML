#### Inheritance in Python

Inheritence is a fundamental concept in Object-Oriented Programming(OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating hot to create and use them in python

```python
## Single Inheritance
## Parent class
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car.")
```

```python
## objects for parent class
car1 = Car(4, 4, "petrol")
car1.drive()
car2 = Car(6, 4, "diesel")  
car2.drive()
```

```python
## Child class
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, autopilot):
        super().__init__(windows, doors, enginetype)
        self.autopilot = autopilot
    
    def selfdrive(self):
        print(f"The person will drive the {self.enginetype} Tesla with autopilot: {self.autopilot}")
```

```python
## object of child class
tesla1 = Tesla(4, 4, "electric", True)
tesla1.drive()
tesla1.selfdrive()
```

```python
# Multiple Inheritance
'''When a class inherits from more than one base class, it is called multiple inheritance. In Python, you can define 
a class that inherits from multiple classes by specifying them in parentheses after the class name.'''
## Base class 1 
class Animal:
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        print("Subclass must impement this method")    

## Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner
        
## Derived class
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)
        
    def speak(self):
        return f"{self.name} says Woof!"
        
# objects of derived class
dog1 = Dog("Buddy", "Alex")    
print(dog1.speak())
print(f"{dog1.name} is owned by {dog1.owner}")
```
