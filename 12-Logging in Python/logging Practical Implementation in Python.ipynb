#### Python Logging

Logging is a crucial of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a fleexible framework for emitting log messages from Python programs. In this lesson, we will cover the basics of logging, including how to configure logging, log levels, and est practices for using logging in python applications.

```python
import logging

## configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

## Logging messages with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

#### Log Levels

Python's logging modules has several log levels indicating the severity of events. The default levels are:

DEBUG: Deatailed information, typically of interest only when diagnosing problems.

INFO: Confimation that things are working as expected.

WARNING: An indication that something unexpected happened or indicative of some problem in the near future (e.g., 'disk space low'). THe software is still working as expected.

ERROR: Due to a more serious problem, the software has not been able to perform some function.

CRITICAL: A very serious erroe, indicating that the program itself may be unable to continue running.

```python
import logging

## configuration logging
logging.basicConfig(
    filename='app.log',  # Log messages will be written to this file
    filemode='w',        # 'w' for overwrite, 'a' for append
    level=logging.DEBUG, # Set the logging level
    format='%(asctime)s- %(name)s - %(levelname)s - %(message)s',  # Log message format
    datefmt='%Y-%m-%d %H:%M:%S'  # Date format in log messages
)

## Logging messages with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```
