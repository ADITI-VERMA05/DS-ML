#### Flask Framework

1. WSGI
2. Jinja2

**FLASK** Web Framework which is created with the help of python programming language.

*Web Server*: where we deploy our enire application. It may be available in AWS EC2, AZURE App, APACHE, IIS. Different web servers are required as per the requirements.

*Web Application*: is created using FLASK Framework and deployed on one of the web server


1. **WSGI**: Web Server Gateway Interface: protocol used by web server to send user_request to web app which in turn generate response to the web server.
2. **Jinja 2**: Web Template Engine: Combine web template/pages with a data source[sql db, ml model, csv sheet, mongodb]. From this data source data we will bw getting this data into the template and this entire web page will get loaded.
