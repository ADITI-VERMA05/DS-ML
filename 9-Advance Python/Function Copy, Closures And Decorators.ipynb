#### Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
#### function copy, closures and decorators
```

```python
## function copy 
def welcome():
    return "Welcome to the world of Python"

welcome()
```

```python
wel = welcome
wel()
```

```python
del welcome
wel()
```

```python
## closures ##
def main_welcome():
    msg = "Welcome"
    def sub_welcome():
        print("Welcome to the advance python course")
        print(msg)
        print("Please enjoy the course")
    return sub_welcome()
```

```python
main_welcome()
```

```python
## closures 1
def main_welcome(func1):
    
    def sub_welcome():
        print("Welcome to the advance python course")
        func1()
        func1("Welcome to the world of Python")
        print("Please enjoy the course")
    return sub_welcome()
```

```python
main_welcome(print) ## print is a function that prints the message passed to it
```

```python
## closures 2
def main_welcome(func2,lst):
    
    def sub_welcome():
        print("Welcome to the advance python course")
        print(func2(lst))
        print("Please enjoy the course")
    return sub_welcome()
```

```python

main_welcome(len,([1,2,3,4,5,6])) ## len is a function that returns the length of the string passed to it
```

```python
## decorators
def main_welcome(func):
    
    def sub_welcome():
        print("Welcome to the advance python course")
        func()
        print("Please enjoy the course")
    return sub_welcome()
```

```python
def course_intro():
    print("This is a course on advance python") 
    
course_intro()
```

```python
main_welcome(course_intro) ## course_intro is a function that prints the course introduction message
```

```python
## decorators ##
@main_welcome
def course_intro():  # Course introduction function decorated with main_welcome
    print("This is a course on advance python") 
```

```python
## decorators 2 

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper  
```

```python
@my_decorator
def say_hello():  ## say_hello function decorated with my_decorator
    print("Hello!")
```

```python
say_hello()  # This will now include the additional behavior defined in the decorator
```

```python
## decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():  ## say_hello function decorated with repeat
    print("Hello!")
```

```python
say_hello()  # This will now include the additional behavior defined in the decorator
```

#### Conclusion

Decoratores are a powerful tool in python for extending and modifying the behaviour of the functions and methods. THey provide a clean and readable way to add functionality such as logging, timing, access control, and more without changing the original code.
