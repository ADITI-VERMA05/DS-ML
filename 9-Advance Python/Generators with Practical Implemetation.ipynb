#### Generators

Generators are a simple way to create iterators. They use the Yield keyword to produce a series of values lazily. which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(5):
        return i * i
```

```python
square(5)
```

```python
## Generators
def square(n):
    for i in range(5):
        yield i * i
```

```python
## Method 1 to create the generator
square(5)
```

```python
for i in square(5):
    print(i)
```

```python
## Method 2 to create the generator
a = square(5)
a
```

```python
## Iterating through this generator
next(a)  # Get the next value from the generator
next(a)  # Get the next value from the generator
```

```python
for i in a:
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_generator()
gen
```

```python
next(gen)  # Get the next value from the generator
```

```python
next(gen)  # Get the next value from the generator
```

```python
for val in gen:
    print(val)
```

### Practical Implementation: Reading Large Files

Practical Implementation: Reading Large Files
Generators are particularly useful for reading large files line by line without loading the entire file into memory.

```python
#### Practical Implementation: Reading Large Files

def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'D:\Krish Naik\DS-ML\9-Advance Python\Large_File.txt' 
for line in read_large_file(file_path):
    print(line.strip())  # Process each line as needed
```
