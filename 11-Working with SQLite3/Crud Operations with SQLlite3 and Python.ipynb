#### SQL and SQLite

SQL (Structured Query Language) is a standard language for managing and manipulating relational databases.
SqLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creatig databases, tables, and performing various SQL operations.

```python
import sqlite3
```

```python
## Connect to an SQLite database
connection=sqlite3.connect('example.db')
connection
```

```python
## creating a cursor object for db to iterate through it
cursor = connection.cursor()
```

```python
## create a table
cursor.execute('''
CREATE TABLE IF NOT EXISTS employee (
    id INTEGER PRIMARY KEY, 
    name Text Not Null,
    age INTEGER,
    department TEXT
    )
''')  

## comit the changes 
connection.commit()
```

```python
cursor.execute('''
select * from employee
''')
```

```python
## Insert data into the sqlite table
cursor.execute('''
INSERT INTO employee (name, age, department)
VALUES ('Krish', 32, 'Data Scientist')
''')

cursor.execute('''
INSERT INTO employee (name, age, department)
VALUES ('Bob', 25, 'Engineering')
''')

cursor.execute('''
INSERT INTO employee (name, age, department)
VALUES ('Charlie', 35, 'Finance')
''')

## commit the changes
connection.commit()
```

```python
## Query the data from the table
cursor.execute('Select * from employee')
rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## Update data into Table
cursor.execute('''
UPDATE employee
set age = 34
where name = "Krish"
''')

connection.commit()
```

```python
## dalete the data from Table
cursor.execute('''
Delete from employee
where name = "Bob"
''')

connection.commit()
```

```python
## Query the data from the table
cursor.execute('Select * from employee')
rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## Working with Sales_data
## Connect to an SQLite database

connection=sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# Create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales (
    Id INTEGER PRIMARY KEY,
    Date TEXT Not NULL,
    Product TEXT NOT NULL,
    Sales INTEGER,
    region TEXT)
''')

## Insert data into sales table
sales_data = [
    ('2023-01-01', 'Product A', 100, 'North'),
    ('2023-01-02', 'Product B', 150, 'South'),
    ('2023-01-03', 'Product C', 200, 'East'),
    ('2023-01-04', 'Product D', 250, 'West'),
    ('2023-01-05', 'Product E', 300, 'North')
]

# Commit
connection.commit()
```

```python
# Insert multiple rows into the sales table [Bulk Insertion]
cursor.executemany('''
INSERT INTO sales (Date, Product, Sales, region)
VALUES (?, ?, ?, ?)
''', sales_data)

# Commit the changes
connection.commit()
```

```python
## Query data from the sales table
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()

#print the queried data
for row in rows:
    print(row)
```

```python
## close the connection
connection.close()
```
